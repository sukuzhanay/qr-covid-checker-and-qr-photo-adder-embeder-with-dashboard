```python
import cv2

cam = cv2.VideoCapture(0)
cv2.namedWindow("Capture la foto con espacio")
img_counter = 0
while True:
    ret, frame = cam.read()
    if not ret:
        print("failed to grab frame")
        break
    cv2.imshow("Capture la foto con espacio", frame)
    k = cv2.waitKey(1)
    if k%256 == 27:
        # ESC pressed
        print("Escape hit, closing...")
        break
    elif k%256 == 32:
        # SPACE pressed
        img_name = "fotoPersona.png"
        cv2.imwrite(img_name, frame)
        print("fotoPersona.png written!")
        break
cam.release()
cv2.destroyAllWindows()
```
