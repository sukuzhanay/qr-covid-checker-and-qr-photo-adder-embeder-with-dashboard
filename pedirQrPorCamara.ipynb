```python
import cv2
from pyzbar.pyzbar import decode

cap = cv2.VideoCapture(0)
cap.set(3, 640)
cap.set(4, 480)
a=""
while True:
    succes, img = cap.read()
    
    for barcode in decode(img):
        myData = barcode.data.decode('utf-8')
        a=myData
    cv2.imshow('result', img)
    cv2.waitKey(1)
    if a:
        break

cap.release()
cv2.destroyAllWindows()
print(a)
    #print(myData)
    
```
